# Properties of String 


Properties of string

- Ordered sequence 
- immutable 
- iterable 
- slicing 
- unicode 
- concatination 
- repetation 


```python
# immutable string
str3 = "This is the third code of string."
print(str3)  # Print the string
print(type(str3))  # Print the type of str3
```

```python
# triple string
str3 = '''This is the third code of string.'''  
print(str3)  # Print the string
```

```python
# Iterate of string
str4 = "This is the fourth code of string." 
for char in str4:
    print(char, end = "")  # Print each character in the string
```

```python
#Slicing of string
str5 = "This is the fifth code of string."
print(str5[0:4])  # Print characters from index 0 to 3
print(str5[5:])  # Print characters from index 5 to the end
```

```python
# unicode string
str6 = u"This is the sixth code of string." 
print(str6)  # Print the string 
```
