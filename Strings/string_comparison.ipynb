# String comparison


- comparison is based on unicode points ( the number assigned to each charater).
- of all character match but length differ , the shorter string is smaller. 

```python
# Lexicographically

str1 = "banana"
str2 = "banned"

print(ord(str1[0]))  # Output: 98 (ASCII value of 'b')
print(ord(str2[0]))  # Output: 98 (ASCII value of 'b')
```

```python
str1 > str2
```

```python
str3 = "zoo"
str4 = "Zoo"

str3 > str4
```

```python
str3 == str4
```

# Coding problem

```python
# count vowels and consonents in a string

s = "_ hel lo 123@world"

for ch in s:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            print(f"{ch} is a vowel")
        else:
            print(f"{ch} is a consonant")
```
