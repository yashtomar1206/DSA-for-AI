# Common String Function 

Function
- lower case 
- upper case 
- strip
- replace 
- split
- join 

```python
# normal string syntax 
str = "HEllo , WOrld"
print(str)
```

```python
# lower case 

str_lower = str.lower()  # Convert the string to lower case
print(str_lower)  # Print the lower case string
```

```python
# upper case 
str_upper = str.upper()  # Convert the string to upper case
print(str_upper)  # Print the upper case string
```

```python
# strip whitespace
str_with_whitespace = "   Hello, World!   "
print(str_with_whitespace.strip())  # Remove leading and trailing whitespace
```

```python
#replace substring
str_replace = str.replace("HEllo", "Hi")  # Replace "HEllo" with "Hi"
print(str_replace)  # Print the modified string
```

```python
# split string
string = "Hello, World! Welcome to Python programming."
split_string = string.split('to')  # Split the string into a list of words
print(split_string)  # Print the list of words
```

```python
# join string
list_of_strings = ["Hello", "World", "Welcome", "to", "Python"]
joined_string = " ".join(list_of_strings)  # Join the list of strings with a space separator
print(joined_string)  # Print the joined string
```
