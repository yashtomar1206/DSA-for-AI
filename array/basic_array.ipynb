# 1. Basic Array Operation 


---- These are the operation of Array 
Traversal 
Insertion 
Deletion 
Access
Searching 
Updating 
Sorting 
Merging 
Splitting 


```python
# creation 
import array as arr

# access 
my_array = arr.array('i', [1, 2, 3, 4, 5])
print(my_array[0])  # Access the first element
print(my_array[2])  # Access the third element
```

Using loop

```python
arr1 = arr.array('i', [1,2,3,4,5,6])

for element in arr1:
    print(element , end = " ")  # Print each element in the array   
```

Insertion 


```python
arr_insert = arr1.pop(5)  # Remove element at index 5
print(arr_insert)  # Print the removed element
print(arr1)  # Print the updated array
```

Deletion

```python
arr2 = arr.array('i', [10, 20, 30, 40, 50])
del arr2[3]  # Remove element at index 3 with the help of del keyword
print(arr2)  # Print the updated array after deletion
```

Updation 


```python
arr2[2]= 100  # Update the element at index 2
print(arr2)  # Print the updated array after modification
```

```python
arr3 = arr.array('i', [5, 2, 9, 1, 5, 6])
arr4= sorted(arr3)  # Sort the array in ascending order
print(arr4)  # Print the sorted array
```

Merging


```python
arr5 = arr1 + arr2  # Concatenate arr1 and arr2
print(arr5)  # Print the concatenated array
```
