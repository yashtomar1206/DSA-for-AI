# 1. Finding the largest number in a list 

```python
numbers = [1, 2, 3, 4, 5,100]

largest = numbers[0] # Initialize largest with the first element

for element in numbers[1:]:
    if element > largest:
        largest = element # Update largest with the new largest element
print("Current largest:", largest)  # Print the current largest element
```

# 2. Count Occurrenceof an element 

```python
items = ["apple", "banana", "cherry", "apple", "orange", "apple"]
target = "apple"

# write code here 

count = 0  # Initialize count to zero
for item in items:
    if item == target:
        count += 1  # Increment count for each occurrence of the target
print("Count of", target, ":", count)  # Print the final count of the target item
```

# 3. Remove duplicate for a List 

```python
numbers = [1, 2, 3, 1, 1, 6,7,9,1]

unique_elements = []  # Initialize an empty list to store unique elements
for number in numbers:
    if number not in unique_elements:
        unique_elements.append(number)  # Add the number to unique_elements if it's not already present
print("Unique elements:", unique_elements)  # Print the list of unique elements
```

# Reverse a List 

```python
data = [4, 5, 6, 2, 8, 2, 4, 6, 8]

n = len(data)  # Get the length of the data list
reversed_data = []  # Initialize an empty list to store reversed elements
for i in range(n-1,-1,-1):  # Iterate backwards through the list
    reversed_data.append(data[i])  # Get the element at the current index
print("Reversed data:", reversed_data)  # Print the reversed list after each addition
```
